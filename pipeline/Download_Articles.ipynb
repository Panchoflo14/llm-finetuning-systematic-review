# <font color='264CC7'>Download_Articles</font>

En este notebook, se genera un script para abrir las páginas de los artículos y descargar los PDFs. La descarga se debe realizar de manera manual, ya que cada página tiene un formato diferente y no es posible automatizar la descarga. Sin embargo, el script permite abrir las páginas de los artículos de manera rápida y sencilla, lo que facilita el proceso de descarga manual.

## <font color='264CC7'>Paquetes y configuración</font>

```python
# Importando librerías para manejo de datos
import pandas as pd
# Librerías auxiliares
import os
import re
```

### <font color='264CC7'>Funciones</font>

```python
def clean_title(title, max_length=100):	
    chars_to_remove = r'[\/:?¿¡!;,\.\'"<>|\*\n\r\[\]]'
    clean_title = re.sub(chars_to_remove, '', title)
    if len(clean_title) > max_length:
        clean_title = clean_title[:max_length]
    return clean_title
```

## <font color='264CC7'>Lectura de datos</font>

```python
# Lectura de archivo xls
df = pd.read_excel("data/articles_selected.xls")

# Tomo los registros de status "Accepted"
df = df[df["status"] == "Accepted"]

# Muestro el tamaño y los primeros 5 registros
print("Total", df.shape[0])
display(df[["bibtex_key", "title","doi"]].head(5))
```

## <font color='264CC7'>Código</font>

```python
# Iteración (este es el que se debe ir cambiando cada vez que se reinicie el proceso)
iteración = 0
# Periodo (este no se cambia)
num_documentos = 20

# tomo los registros a partir de periodo
df_temp = df.iloc[num_documentos*iteración:num_documentos*(iteración+1)]

# inicio en la fila = periodo
for index, row in df_temp.iterrows():
    id = row["bibtex_key"]
    paper = row["doi"]
    paper = "https://doi.org/"+paper
    title = clean_title(row["title"])

    os.system(f"start microsoft-edge:{paper}")
    print(f"Abriendo {id} - {title}")
```
